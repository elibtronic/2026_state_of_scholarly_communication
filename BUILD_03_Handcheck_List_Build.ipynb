# Handcheck List Build

Will find out which articles have Brock authors as correponding and will try to kick out a CSV of titles to review.

```python
import pandas as pd
import numpy as np
import pickle
```

```python
#pd.set_option('display.max_columns', None)
#pd.set_option('display.max_rows', None)
#pd.set_option('display.max_colwidth', 600)
CLOSED_WORKS_ANALYZED_PICKLE = "2026_CLOSED_ANALYZED.pkl"
CHECK_CSV_FILE = "2026_Papers_to_Check.csv"
```

```python
with open(CLOSED_WORKS_ANALYZED_PICKLE, 'rb') as file:
    works = pickle.load(file)
```

```python
#free_green = True
works = works[works['free_green']==True]
works.reset_index(inplace=True)
del(works['level_0'])
del(works['index'])


a_name = []
for i, r in works.iterrows():

    for a in r['authorships']:
        if a['author_position'] == 'first':
            if "Brock" in str(a['raw_affiliation_strings'][0]):
                #print(a['author']['display_name'])
                #print(a['raw_affiliation_strings'][0])
                a_name.append(a['author']['display_name'])
                break
        else:
            a_name.append(np.nan)
            break

works['Brock_Primary_Author'] = a_name
```

```python
papers_to_check = works[works['Brock_Primary_Author'].notna()]
papers_to_check.reset_index(inplace=True)
```

```python

syp_links = []
stub = "https://researchguides.library.brocku.ca/getting-your-research-out/green-open-access?doi="

for i,r in papers_to_check.iterrows():
    syp_links.append(stub+r['doi'].split("https://doi.org/")[1])

papers_to_check['syp'] = syp_links
```

```python
keep = [np.nan]*len(papers_to_check)
papers_to_check['keep'] = keep
```

```python
#This will be the hand check list
# - Check to make sure article is a full-text things we want a copy of
# - Check to make sure SYP feedback is good

# How do I encode the results of this check in a column?

papers_to_check.to_csv(CHECK_CSV_FILE,index=False)
papers_to_check[['doi','Brock_Primary_Author','syp']]
```

## Checking procedure

- check the DOI to make sure it is a keeper
- check the SYP link to make sure it shows a good result
- Keep = True for good ones
- Keep = Fals for ones to toss
